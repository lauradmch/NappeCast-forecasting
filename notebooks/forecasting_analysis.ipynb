## Holt-Winters exponential smoothing 
(= triple exponential smoothing / ETS with additive trend+seasonality) -> forecasting method that decomposes a series into level, trend, and seasonal components, each updated recursively with its own smoothing weight.

The three smoothing weights to watch (fitted by MLE minimizing SSE):

- smoothing_level (α = 0.989) = how much weight the level update gives to the newest observation vs. the model's own extrapolation. α near 1 means the level barely smooths at all; it chases the latest point almost exactly. That's unusually high and worth flagging: it suggests either genuine high-frequency variability in your groundwater level, or the optimizer overfitting to noise.
- smoothing_trend (β = 0.228) = analogous weight for updating the trend (slope) estimate. Moderate value, trend adapts at a middling pace.
- smoothing_seasonal (γ = 0.011) = weight for updating the weekly (period=7) seasonal pattern. Essentially zero: the seasonal shape barely evolves after initialization, it stays close to initial_seasons for the whole series.

The initial/state values:

- initial_level = 102.02, initial_trend ≈ 0.012 (per day) — starting level and slope, estimated heuristically (initialization_method="heuristic") from the early part of the series, not optimized jointly with α/β/γ.
- initial_seasons — 7 values, one per weekday, the additive offset from the level. Magnitudes are ~0.001–0.02, negligible next to a level of ~102. Your seasonal component is essentially flat — there's little real weekly cycle in groundwater level, which is physically sensible (nappe phréatique dynamics are driven by rainfall/infiltration, not weekday effects).
- damping_trend = NaN, use_boxcox = False, lamda = None, remove_bias = False — these are all inactive options: no damped trend, no Box-Cox variance-stabilizing transform, no bias correction on the forecast.

Observation: Holt-Winters only supports one seasonal period at a time -> not the right model here: Holt-Winters has to store and update a separate seasonal value for each of the 365 days of the cycle, all sharing one smoothing weight (γ), with only a handful of years of data that's too many parameters to estimate reliably, and it treats day 45 and day 46 as unrelated numbers rather than points on a smooth curve. It also can't combine that yearly cycle with an outside driver like rainfall.

## STL decomposition 
(STL(series, period=365)) ->
 Explicitly pull apart trend/seasonal/residual, inspect the annual component visually, then model trend+residual with ARIMA on the deseasonalized residual.

```python
from statsmodels.tsa.forecasting.stl import STLForecast
from statsmodels.tsa.arima.model import ARIMA

# STLForecast: decomposes `series` with STL (period=365, same yearly
# assumption as before), fits `model` only on the deseasonalized part
# (trend + noise — no 365-parameter seasonal fit needed), then forecasts
# and adds the seasonal component back in automatically.
stlf = STLForecast(
    series,
    ARIMA, # ARIMA (AutoRegressive Integrated Moving Average) = statistical time series model
    model_kwargs=dict(order=(1, 1, 1), trend="n"),
    period=365,
)
stlf_res = stlf.fit()

y_pred = stlf_res.forecast(90)

plt.figure(figsize=(14, 6))
plt.plot(series["2025-01-01":], label="Observed groundwater level")
plt.plot(y_pred, label="Forecast (STL + ARIMA)")
plt.title("Groundwater level forecast (STL decomposition + ARIMA)")
plt.xlabel("Date")
plt.ylabel("Groundwater level (m)")
plt.legend()
plt.show()
```

## SARIMAX
The highest-leverage improvement, though, is probably not about the seasonal period at all: groundwater level is physically driven by rainfall/recharge and extraction, not the calendar. If you have or can source daily/monthly precipitation (easy to get — e.g. Météo-France, or an open weather API) or pumping data for your site, feeding it as an exogenous regressor (SARIMAX(series, exog=rainfall, order=..., seasonal_order=...), or as a regressor in Prophet/UnobservedComponents) will very likely outperform any purely calendar-based seasonality.
- Before comparing models: switch from eyeballing one 90-day forecast to walk-forward backtesting (sklearn.model_selection.TimeSeriesSplit or rolling-origin manually) and compare RMSE/MAE out-of-sample — in-sample fit with α≈0.99 is not a reliable guide to which model will actually forecast better.
